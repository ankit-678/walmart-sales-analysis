```python
#missing value
df.isnull().sum()
```

```python
#all duplicates removes
df.drop_duplicates(inplace=True)
```

```python
#deopping all rows with missing records
df.dropna(inplace=True)
#verify
df.isnull().sum()
```

```python
#convert datatypes (unit-price )
df.dtypes
```

```python
# replace the unit-price dollar 
df['unit_price']=df['unit_price'].str.replace('$','').astype(float)
df.head()
```

```python
df.columns
```

```python
df.columns=df.columns.str.lower()
```

```python
#total calculated
df['total']= df['unit_price'] * df['quantity']
df.head()
```
